```python
from deepface import DeepFace
import os

carpeta = "/Users/dani/INTELIGENCIA ARTIFICIAL 4TO CUATRI/PROYECTO_LOGEO_RECONOCIMIENTO_FACIAL/notebooks/caras_registradas"
login_image = "/Users/dani/INTELIGENCIA ARTIFICIAL 4TO CUATRI/PROYECTO_LOGEO_RECONOCIMIENTO_FACIAL/notebooks/caras_registradas/login_attempt.jpg"

acceso_concedido = False

for archivo in os.listdir(carpeta):

    # Ignorar archivos ocultos o carpetas
    if not archivo.lower().endswith((".jpg", ".jpeg", ".png")):
        continue

    ruta_usuario = os.path.join(carpeta, archivo)

    try:
        resultado = DeepFace.verify(
            img1_path=login_image,
            img2_path=ruta_usuario,
            detector_backend="retinaface"
        )

        if resultado["verified"]:

            print("\nAcceso concedido")
            print(f"Usuario identificado: {archivo}")

            acceso_concedido = True
            break

    except Exception as e:
        print(f"Error con {archivo}: {e}")

if not acceso_concedido:
    print("\nAcceso denegado")
```
